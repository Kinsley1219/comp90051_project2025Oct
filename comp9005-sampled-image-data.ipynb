```python
import os

base_dir = "/kaggle/input/celeba-dataset"
print(os.listdir(base_dir))
print(os.listdir(os.path.join(base_dir, "img_align_celeba"))[:10]) 
```

```python
import os, random, shutil, zipfile

src_dir = "/kaggle/input/celeba-dataset/img_align_celeba/img_align_celeba"
dst_dir = "/kaggle/working/celeba_subset"
os.makedirs(dst_dir, exist_ok=True)

all_imgs = os.listdir(src_dir)
sample_imgs = random.sample(all_imgs, 12000)

for fname in sample_imgs:
    shutil.copy(os.path.join(src_dir, fname), os.path.join(dst_dir, fname))

print(f"The subset has been generated {len(os.listdir(dst_dir))} images")

zip_path = "/kaggle/working/celeba_subset.zip"
with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for fname in os.listdir(dst_dir):
        zipf.write(os.path.join(dst_dir, fname), arcname=fname)

print("celeba_subset.zip，downloaded from the Output ")
```
